```python
import os
import pandas as pd

import requests
import googlemaps

from geopy.distance import distance
from geopy.distance import geodesic

from dotenv import load_dotenv

from tqdm import tqdm
```

```python
# Load environment variables
load_dotenv()
gmap_api_key =  os.getenv('GMAP_API_KEY')
```

```python
gmap_client_key = gmap_api_key
gmaps = googlemaps.Client(key=gmap_client_key)
```

```python
ma_center_coords = (42.4072, -71.3824) 

# Google Maps API handler
def callOG_GoogleMapsAPI(location):
    try:
        # Locations are limited to Massachusetts for now
        geocode_result = gmaps.geocode(f"{location}, Massachussetts", components={"administrative_area_level": "MA", "country": "US"})
        
        if (len(geocode_result) > 0):
            longitude = geocode_result[0]['geometry']['location']['lng']
            latitude = geocode_result[0]['geometry']['location']['lat']
            coords = [latitude, longitude]

            # Check if location couldn't be identified (it returns the center of MA)
            if geodesic(coords, ma_center_coords).km <= 10:
                print(f"[INFO] Location {location} could not be found in MA")
                return [None, None]
            else:
                return coords
        else:
            print(f"[WARNING] Could not find location {location} through OG google maps")
            return [None, None]
    except Exception as error:
        print(f"[ERROR] Error finding locations through google maps, {error}")
        return [None, None]
```

```python
# Google Maps API handler
def callGoogleMapsAPI(location):
    try:
        # Locations are limited to Massachusetts for now
        geocode_result = gmaps.geocode(f"{location}", components={"country": "US"})
        
        if (len(geocode_result) > 0):
            longitude = geocode_result[0]['geometry']['location']['lng']
            latitude = geocode_result[0]['geometry']['location']['lat']

            # Get the city from the address components
            address_components = geocode_result[0]['address_components']
            city = None
            for component in address_components:
                if 'locality' in component['types']:
                    city = component['long_name']
                    break
            
            return [latitude, longitude], city
        else:
            print(f"[WARNING] Could not find location {location} through google maps")
            return [None, None], None
    except Exception as error:
        print(f"[ERROR] Error finding locations through google maps, {error}")
        return [None, None], None
```

```python
from pymongo import MongoClient
from dotenv import load_dotenv
import os

load_dotenv()
collection_name = "articles_data"
mongo_uri = os.getenv("MONGO_URI_NAACP")
mongo_db_name = os.getenv("MONGO_DB_NAME_NAACP")
mongo_client = MongoClient(mongo_uri)

# Set up MongoDB connection
db = mongo_client[mongo_db_name]

location_collection = db["locations_data"] 

locations = list(location_collection.find())
```

```python
reduced_locations = locations
rows = []
for location in tqdm(reduced_locations):
    try:
        name = location["value"]
        coords = location["coordinates"]
        coords = [coords[1], coords[0]]

        general_coords, new_city = callGoogleMapsAPI(name)
        
        if coords[0] is not None and general_coords[0] is not None:
            distance_km = distance(coords, general_coords).km
        else:
            distance_km = None

        rows.append({
            "location": name,
            "city": location["city"],
            "new_city": new_city,
            "coords": coords,
            "general_coords": general_coords,
            "distance_km": distance_km
        })
    except Exception as error:
        print(f"[ERROR] Error processing location {location['value']}, {error}")
        continue

locations_df = pd.DataFrame(rows)
```

```python
def filter_large_distances(df, threshold=100):
    # Filter the DataFrame for rows where the Distance is greater than the threshold
    large_distance_df = df[df['Distance'] > threshold]

    return large_distance_df
```

```python
# Define Massachusetts latitude and longitude boundaries
ma_lat_bounds = (42, 42.8)  # Latitude range for MA
ma_lng_bounds = (-73.50, -70.0)  # Longitude range for MA

# Function to check if coordinates are within MA
def is_within_ma(coords):
    lat, lng = coords
    return ma_lat_bounds[0] <= lat <= ma_lat_bounds[1] and ma_lng_bounds[0] <= lng <= ma_lng_bounds[1]

# Apply the function to check if OG_Coordinates and Free_Coordinates are within MA
locations_df['OG_in_MA'] = locations_df['OG_Coordinates'].apply(lambda x: is_within_ma(x) if x[0] is not None else False)
locations_df['Free_in_MA'] = locations_df['Free_Coordinates'].apply(lambda x: is_within_ma(x) if x[0] is not None else False)

# Count how many are within MA
og_in_ma_count = locations_df['OG_in_MA'].sum()
free_in_ma_count = locations_df['Free_in_MA'].sum()

print(f"Number of original coordinates within MA: {og_in_ma_count}")
print(f"Number of free coordinates within MA: {free_in_ma_count}")
```

```python
import folium


def visualize_large_distances_on_map(df, threshold=100, output_file='large_distances_map.html'):
    # Filter the DataFrame for rows where the Distance is greater than the threshold
    large_distance_df = df[df['Distance'] > threshold]
    
    # Initialize a folium map centered around Massachusetts (you can adjust the center as needed)
    m = folium.Map(location=[42.4072, -71.3824], zoom_start=7)
    
    # Add markers to the map
    for _, row in large_distance_df.iterrows():
        location = row['location']
        og_coords = row['OG_Coordinates']
        free_coords = row['Free_Coordinates']
        distance_km = row['Distance']

        # Add a marker for the original coordinates
        if og_coords and og_coords[0] is not None:
            folium.Marker(
                location=og_coords,
                popup=f"Location: {location}\nOG Coordinates: {og_coords}\nDistance: {distance_km:.2f} km",
                icon=folium.Icon(color='blue', icon='info-sign')
            ).add_to(m)
        
        # Add a marker for the free coordinates
        if free_coords and free_coords[0] is not None:
            folium.Marker(
                location=free_coords,
                popup=f"Location: {location}\nFree Coordinates: {free_coords}\nDistance: {distance_km:.2f} km",
                icon=folium.Icon(color='red', icon='info-sign')
            ).add_to(m)
        
        # Optionally, draw a line between the original and free coordinates
        if og_coords and free_coords and og_coords[0] is not None and free_coords[0] is not None:
            folium.PolyLine(
                [og_coords, free_coords],
                color='green',
                weight=2,
                opacity=0.6
            ).add_to(m)

    # Save the map to an HTML file
    m.save(output_file)
    print(f"Map with large distances saved as {output_file}")

    return m
```
